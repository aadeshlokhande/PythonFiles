# Importing Numpy

```python
import numpy as np
```

### Basic Array Creation

```python
arr = np.array([1,2,3,4])
arr
```

```python
lst = [1,2,3,4,5, "hello"]
lst
```

```python
np.array(lst)
```

> data store in float values

```python
a = [12,3,4,5,3,43.4]
ar = np.array(a)
ar
```

> data stored in multi dimension

```python
l = [[1,2,3],[4,5,6],[7,8,9]]
arr = np.array(l)
arr
```

## Array Generation Functions

```python
np.arange(1,11) # array 1 to 10
```

```python
np.zeros(10)   # get 10 zeros
```

```python
np.zeros((3,5))
```

```python
np.ones(6)
```

```python
np.ones((3,5))
```

```python
np.linspace(1,5,10) # linear space between sp to ep
```

## Random Number Generation

```python
np.random.rand(5)
```

```python
np.random.rand(3,5) # numbers between 0 to 1
```

```python
np.random.randn(3,3) # normalization
```
